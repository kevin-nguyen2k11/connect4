```python
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime
import tensorflow as tf
from tensorflow import keras
from keras import layers
from keras import regularizers
import logging
import pickle
import select
import threading
```

```python
class Config:
    def __init__(self):
        #Game specific settings
        self.width=7
        self.height=6
        self.win_length=4
        self.max_moves=self.width*self.height
        self.dirs=[(1,0),(1,1),(0,1),(-1,1)]
        
        #Self play
        self.num_saved_games=10000
        self.num_simultaneous_games=8
        self.num_high_temp_moves=5
        self.num_simulations=70
        self.num_threads=10
        self.buffer_directory='/Users/kevinnguyen/Documents/Python files/connect_4/data'

        #Root prior exploration noise
        self.alpha=0.1
        self.epsilon=0.25

        #UCB formula
        self.c_base=19652
        self.c_init=1.15
        
        #Evaluation
        self.num_eval_games=100
        self.win_margin=0.55

        #Training
        # self.num_workers=10-self.num_simultaneous_games-2
        self.num_workers=1
        self.batch_size=128
        self.initial_epoch=770
        self.epochs_per_update=5
        self.epochs_per_checkpoint=300
        self.batches_per_epoch=50
        self.momentum=0.9
        self.model_directory='/Users/kevinnguyen/Documents/Python files/connect_4/models'
        
        self.learning_rate_schedule={
            0:0.0002, #0.002
            1500:0.00002,
            30000:0.000002,
            50000:0.0000002
        }
```

```python
class GameController:
    def __init__(self,config,num_moves=None,ref_board=None,board=None):
        self.config=config
        self.width=config.width
        self.height=config.height
        self.win_length=config.win_length
        self.max_moves=config.max_moves
        self.dirs=np.asarray(config.dirs)
        self.num_moves=num_moves or 0
        self.move_history=np.empty((2,self.max_moves),dtype=int)
        self.policy_history=[]
        self.win_state=None
        self.current_player=self.num_moves%2
        self.ref_board=np.zeros((self.height,self.width,2),dtype=int) if ref_board is None else ref_board
        self.board=np.zeros((4,self.height,self.width,2),dtype=int) if board is None else board

    def __str__(self):
        return np.array2string(np.flip(self.ref_board[:,:,0]+self.ref_board[:,:,1]*2,0))    
    
    def clone(self):
        return GameController(self.config,self.num_moves,self.ref_board.copy(),self.board.copy())
    
    def random_reflect(self,state):
        if np.random.randint(2):
            return np.flip(state,axis=1)
        return state
    
    def get_state(self):
        if self.current_player:
            return self.random_reflect(np.flip(self.ref_board,axis=-1))
        else:
            return self.random_reflect(self.ref_board)

    def get_legal_moves(self):
        if self.win_state!=None:
            return np.full(self.width,False)
        top_row=np.sum(self.ref_board[-1,:,:].T,axis=0)
        return top_row==0
    
    def make_move(self,x):
        if self.win_state!=None:
            print('Game already done')
            return None
        if x<0 or x>self.width:
            raise ValueError('Illegal move, chosen column outside of board boundaries')
        col=np.sum(self.ref_board[:,x,:].T,axis=0)
        y=None
        for i in range(len(col)):
            if col[i]==0:
                y=i
                break
        if y==None:
            raise ValueError('Illegal move, chosen column is full')
        self.ref_board[y,x,self.current_player]=1
        self.board[:,y,x,self.current_player]=1
        self.move_history[:,self.num_moves]=[x,y]
        self.num_moves+=1
        if self.is_win_condition(x,y):
            self.delete_boards
            return
        self.current_player^=1

    def is_win_condition(self,x,y):
        current_board=self.board[:,:,:,self.current_player]
        for i,dir in enumerate(self.dirs):
            board_slice=current_board[i]
            dir_tuple=(dir,-1*dir)
            temp=[0]*2
            for j in range(2):
                coord=np.asarray([y,x])+dir_tuple[j]
                if coord[0]>=0 and coord[0]<self.height and coord[1]>=0 and coord[1]<self.width:
                    temp[j]=board_slice[tuple(coord)]
            board_slice[y,x]+=temp[0]+temp[1]
            if board_slice[y,x]>=self.win_length:
                self.win_state=1
                return True
            for j in range(2):
                for k in range(temp[j]):
                    coord=np.asarray([y,x])+(1+k)*dir_tuple[j]
                    board_slice[tuple(coord)]=board_slice[y,x]
        if self.num_moves==self.max_moves:
            self.win_state=0
            return True
        return False
    
    def delete_boards(self):
        self.ref_board=None
        self.board=None
    
    def store_policy(self,policy):
        self.policy_history.append(policy)
    
    def state_from_hist(self,move_ind):
        if move_ind>=self.num_moves:
            raise ValueError('Move number out of range of game length')
        board=np.zeros((self.height,self.width,2),dtype=int)
        player=move_ind%2
        for i in range(2):
            ind=self.move_history[:,player:move_ind+1:2]
            board[:,:,i][ind[1],ind[0]]=1
            player^=1
        return self.random_reflect(board)
    
    def label_from_hist(self,move_ind):
        factor=1 if (move_ind%2)==(self.num_moves%2) else -1            
        return (self.policy_history[move_ind],factor*self.win_state)
```

```python
class Node:
    def __init__(self,move=None,parent=None,c_base=None,c_init=None):
        self.move=move
        self.parent=parent
        self.children={}
        self.c_base=c_base
        self.c_init=c_init
        self.legal_moves=None
        self.child_priors=None
        self.child_visits=None
        self.child_mean_actions=None
        self.lock=threading.Lock()
    
    @property
    def visits(self):
        return self.parent.child_visits[self.move]
    
    @visits.setter
    def visits(self,num_visits):
        self.parent.child_visits[self.move]=num_visits
    
    @property
    def mean_action(self):
        return self.parent.child_mean_actions[self.move]
    
    @mean_action.setter
    def mean_action(self,action_value):
        self.parent.child_mean_actions[self.move]=action_value
    
    def set_legal_moves(self,legal_moves):
        self.legal_moves=legal_moves

    def create_children(self,priors):
        priors=np.exp(priors)
        priors[~self.legal_moves]=np.nan
        if (prior_sum:=np.nansum(priors))!=0:
            priors/=prior_sum
        max_children=np.size(priors)
        self.child_priors=priors
        self.child_visits=np.zeros(max_children)
        self.child_mean_actions=np.zeros(max_children)
        for move,is_legal in enumerate(self.legal_moves):
            if is_legal:
                self.children[move]=Node(move,self,self.c_base,self.c_init)
            
    def update(self,value):
        self.mean_action=((self.visits-1)*self.mean_action+value)/self.visits
        
    def get_appraisal(self):
        c=np.log((self.visits+self.c_base+1)/self.c_base)+self.c_init
        return self.child_mean_actions+c*self.child_priors*(np.sqrt(self.visits)/(1+self.child_visits))
    
    def choose_child(self):
        scores=self.get_appraisal()
        move=np.nanargmax(scores)
        return move
    
    def choose_move(self,temperature):
        if temperature==1:
            pi=np.exp(self.child_visits)
            pi[~self.legal_moves]=0
            pi/=np.sum(pi)
            move=np.random.choice(np.arange(len(pi)),p=pi)
        else:
            visits=np.copy(self.child_visits)
            visits[~self.legal_moves]=np.nan
            move=np.arange(len(visits))[visits==np.nanmax(visits)]
            if len(move)>1:
                scores=self.get_appraisal()
                move=move[np.argmax(scores[move])]
            else:
                move=move[0]
        return move
    
class MonteCarloTree:
    def __init__(self,config,model,noise=True):
        dummy_node=Node()
        dummy_node.child_visits=np.zeros(1)
        dummy_node.child_mean_actions=np.zeros(1)
        dummy_node.children={0:Node(0,dummy_node,config.c_base,config.c_init)}
        self.root=dummy_node.children[0]
        self.model=model
        self.num_simulations=config.num_simulations
        self.num_threads=config.num_threads
        self.queue_states=np.empty((self.num_threads,config.height,config.width,2))
        self.queue_full=np.zeros(self.num_threads,dtype=int)
        self.results={}
        self.queue_condition=threading.Condition()
        self.results_condition=threading.Condition()
        self.noise=noise
        self.alpha=config.alpha
        self.epsilon=config.epsilon

    def choose_move(self,game,temperature): 
        if self.noise and self.root.children:
            self.add_noise()
        evaluator=threading.Thread(target=self.evaluate)
        evaluator.start()
        for i in range(self.num_simulations//self.num_threads):
            threads=[]
            for j in range(self.num_threads):
                sim_game=game.clone()
                threads.append(threading.Thread(target=self.search,args=[self.root,sim_game,j]))
                threads[-1].start()
            for thread in threads:
                thread.join()
        evaluator.join()
        move=self.root.choose_move(temperature)
        pi=self.root.child_visits/np.sum(self.root.child_visits)
        self.root.parent=None
        self.root=self.root.children[move]
        return (move,pi)
        
    def evaluate(self):
        for i in range(self.num_simulations//self.num_threads):
            self.queue_condition.acquire()
            while not np.all(self.queue_full):
                self.queue_condition.wait()
            self.queue_full[:]=0
            self.queue_condition.release()
            
            self.results_condition.acquire()
            priors_list,value_list=self.model(self.queue_states,training=False)
            for j in range(self.num_threads):
                self.results[j]=(priors_list[j].numpy().flatten(),value_list[j].numpy().flatten())
            self.results_condition.notify_all()
            self.results_condition.release()
        
    def search(self,current_node,game,thread_id):
        current_node.lock.acquire()
        current_node.visits+=1
        current_node.lock.release()
        if not current_node.children:
            if current_node.legal_moves is None:
                current_node.legal_moves=game.get_legal_moves()
            self.queue_condition.acquire()
            if np.any(current_node.legal_moves):
                self.queue_states[thread_id,:,:,:]=game.get_state()
            self.queue_full[thread_id]=1
            self.queue_condition.notify()
            self.queue_condition.release()

            self.results_condition.acquire()
            while thread_id not in self.results.keys():
                self.results_condition.wait()
            self.results_condition.release()

            if np.any(current_node.legal_moves):
                priors,value=self.results.pop(thread_id)
                current_node.create_children(priors)
            else:
                del self.results[thread_id]
                value=game.win_state
        else:
            move=current_node.choose_child()
            game.make_move(move)
            value=self.search(current_node.children[move],game,thread_id)
        current_node.lock.acquire()
        current_node.update(value)
        current_node.lock.release()
        return -value
    
    def add_noise(self):
        priors=self.root.child_priors
        self.root.child_priors=(np.full(np.size(priors),(1-self.epsilon))*priors+self.epsilon
                                *np.random.dirichlet(np.full(np.size(priors),self.alpha)))
    
    def opponent_move(self,move):
        if self.root.children:
            self.root.parent=None
            self.root=self.root.children[move]

    def resign(self):
        pass
```

```python
def get_model(config,version):
    directory=config.model_directory
    model=initialize_model(config)
    model.load_weights(directory+'/'+version+'/model').expect_partial()
    return model

def initialize_model(config):
        inputs=keras.Input(shape=(config.height,config.width,2),name='board')
        x=layers.Conv2D(256,3,kernel_regularizer=regularizers.L2(l2=0.0001),
                        kernel_initializer='Orthogonal')(inputs)
        x=layers.BatchNormalization()(x)
        conv_block=layers.ReLU()(x)

        x=layers.Conv2D(256,3,padding='same',kernel_regularizer=regularizers.L2(l2=0.0001),
                        kernel_initializer='Orthogonal')(conv_block)
        x=layers.BatchNormalization()(x)
        x=layers.ReLU()(x)
        x=layers.Conv2D(256,3,padding='same',kernel_regularizer=regularizers.L2(l2=0.0001),
                        kernel_initializer='Orthogonal')(x)
        x=layers.BatchNormalization()(x)
        add=layers.add([conv_block,x])
        x=layers.ReLU()(add)

        #policy head
        policy=layers.Conv2D(2,1,kernel_regularizer=regularizers.L2(l2=0.0001),
                             kernel_initializer='Orthogonal')(x)
        policy=layers.BatchNormalization()(policy)
        policy=layers.ReLU()(policy)
        policy=layers.Flatten()(policy)
        policy_output=layers.Dense(config.width,name='policy',kernel_regularizer=regularizers.L2(l2=0.0001),
                                   kernel_initializer='Orthogonal')(policy)

        #value head
        value=layers.Conv2D(1,1,kernel_regularizer=regularizers.L2(l2=0.0001),
                            kernel_initializer='Orthogonal')(x)
        value=layers.BatchNormalization()(value)
        value=layers.ReLU()(value)
        value=layers.Flatten()(value)
        value=layers.Dense(256,kernel_regularizer=regularizers.L2(l2=0.0001),
                           kernel_initializer='Orthogonal')(value)
        value=layers.ReLU()(value)
        value_output=layers.Dense(1,activation='tanh',name='value',kernel_regularizer=regularizers.L2(l2=0.0001),
                                  kernel_initializer='Orthogonal')(value)

        def custom_loss(y_true,y_pred):
            loss=(tf.keras.losses.mean_squared_error(tf.stop_gradient(y_true[1]),y_pred[1])
                  +tf.nn.softmax_cross_entropy_with_logits(labels=tf.stop_gradient(y_true[0]),logits=y_pred[0]))
            return loss

        avg_loss=tf.keras.metrics.Mean('loss',dtype=tf.float32)

        class custom_fit(tf.keras.Model):
            def train_step(self,data):
                states,labels=data
                with tf.GradientTape() as tape:
                    outputs=self(states,training=True) # forward pass 
                    reg_loss=tf.reduce_sum(self.losses)
                    pred_loss=custom_loss(labels,outputs)
                    total_loss=tf.reduce_sum(pred_loss)+reg_loss
                gradients=tape.gradient(total_loss,self.trainable_variables)
                self.optimizer.apply_gradients(zip(gradients,self.trainable_variables))
                avg_loss.update_state(total_loss)
                return {'loss': avg_loss.result()}

            @property
            def metrics(self):
                return [avg_loss]
        
        class LRSchedule(tf.keras.optimizers.schedules.LearningRateSchedule):
            def __init__(self,config):
                self.learning_rate_schedule=config.learning_rate_schedule
            
            def __call__(self,step):
                lr=None
                for key in self.learning_rate_schedule.keys():
                    if step.numpy()>=key:
                        lr=self.learning_rate_schedule[key]
                return lr
            
        lr_schedule=LRSchedule(config)
        model=custom_fit(inputs=[inputs],outputs=[policy_output,value_output],name='f_theta')
        opt=tf.keras.optimizers.legacy.SGD(learning_rate=lr_schedule,momentum=config.momentum)
        model.compile(optimizer=opt,run_eagerly=True)
        return model
```

```python
def plot_board(board):
      board=np.flip(board[:,:,0]+board[:,:,1]*10,0)
      img=plt.imshow(board,interpolation='none')
      plt.axis('off')
      plt.show()  

class HumanPlayer():
    def __init__(self):
        return
    
    def choose_move(self,game,temperature):
        print('Please make a move:')
        move=int(input())
        return move,None
    
    def opponent_move(self,move):
        return

def human_play(config):
    game=GameController(config)
    model=get_model(config,'best')
    tree=MonteCarloTree(config,model,noise=False) #increase thinking time
    human=HumanPlayer()
    players=[tree,human]
    current_player=np.random.randint(2)
    plot_board(game.ref_board)
    while game.win_state==None:
        move,pi=players[current_player].choose_move(game,0)
        try:
            game.make_move(move)
            current_player^=1
        except ValueError:
            continue
        plot_board(game.ref_board)
        players[current_player].opponent_move(move)
    if game.win_state==0:
        print('Draw!')
    elif not current_player:
        print('You won!')
    else:
        print('Skill issue')

if __name__=='__main__':
    config=Config()
    human_play(config)
```
